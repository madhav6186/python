25. Parking Fee

```python
vehicle = input("Enter vehicle: ")
hours = int(input("Enter hours: "))

if vehicle == "bike":
    if hours <= 2:
        fee = hours * 20
    else:
        fee = 40 + (hours - 2) * 10

elif vehicle == "car":
    if hours <= 2:
        fee = hours * 40
    else:
        fee = 80 + (hours - 2) * 20

elif vehicle == "bus":
    if hours <= 2:
        fee = hours * 60
    else:
        fee = 120 + (hours - 2) * 30

else:
    fee = 0
    print("Invalid vehicle")

print("Parking fee =", fee)
```

28. Password Strength

```python
password = input("Enter password: ")

if len(password) < 6:
    print("Weak")

elif len(password) <= 9:
    print("Medium")

else:
    upper = False
    lower = False
    digit = False

    for x in password:
        if x.isupper():
            upper = True
        if x.islower():
            lower = True
        if x.isdigit():
            digit = True

    if upper and lower and digit:
        print("Strong")
    else:
        print("Medium")
```

30. Smart ATM

```python
balance = int(input("Enter balance: "))
amount = int(input("Enter withdrawal amount: "))
pin = int(input("Enter PIN: "))

if pin != 1234:
    print("Wrong PIN")

elif amount <= 0:
    print("Invalid amount")

elif amount % 100 != 0:
    print("Amount must be a multiple of 100")

elif amount > balance:
    print("Insufficient balance")

elif balance - amount < 500:
    print("Minimum balance should be ₹500")

else:
    balance = balance - amount
    print("Withdrawal successful")
    print("Remaining balance =", balance)
```
