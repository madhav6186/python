8. Convert seconds into hours, minutes and seconds

9. Extract hundreds, tens and units

```python
n = int(input("Enter a 3-digit number: "))

hundreds = n // 100
tens = (n // 10) % 10
units = n % 10

print("Hundreds =", hundreds)
print("Tens =", tens)
print("Units =", units)
```
